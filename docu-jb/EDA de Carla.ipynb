##### 📌 Distribución de la variable objetivo

```python
sns.countplot(data=df, x="Survived")
plt.title("Supervivencia (0 = No, 1 = Sí)")
plt.show()
```

### 🚑 Distribución de Supervivencia

Aquí observamos el desbalance entre pasajeros que sobrevivieron y no sobrevivieron.
Suele haber más fallecidos que supervivientes.

##### 📌 Distribuciones univariantes

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))
sns.histplot(df["Age"].dropna(), bins=30, ax=axes[0])
axes[0].set_title("Distribución de Age")

sns.histplot(df["Fare"], bins=30, ax=axes[1])
axes[1].set_title("Distribución de Fare")
plt.show()
```

### 🎚️ Distribución de variables numéricas

Las distribuciones de `Age` y `Fare` permiten observar:

- `Age` tiene una distribución aproximadamente unimodal.
- `Fare` tiene una cola larga hacia la derecha, indicando algunos pasajeros con tarifas muy elevadas.


##### 📌 Variables categóricas

```python
cat_cols = ["Sex", "Pclass", "Embarked"]
for col in cat_cols:
    sns.countplot(data=df, x=col)
    plt.title(f"Distribución de {col}")
    plt.show()
```

### 🧮 Distribución de variables categóricas

Analizamos las principales variables categóricas:
- **Sex**: repartición por género.
- **Pclass**: categoría socioeconómica (1ª, 2ª, 3ª clase).
- **Embarked**: puerto de embarque.

Estas variables tienen un impacto importante en la tasa de supervivencia.

##### 📌 Relación entre variables y supervivencia

```python
sns.countplot(data=df, x="Sex", hue="Survived")
plt.title("Supervivencia según género")
plt.show()

sns.countplot(data=df, x="Pclass", hue="Survived")
plt.title("Supervivencia según clase")
plt.show()

sns.boxplot(data=df, x="Survived", y="Age")
plt.title("Edad según supervivencia")
plt.show()
```

### 🎯 Variables asociadas a la supervivencia

Aquí observamos:
- **Sexo**: Las mujeres sobrevivieron en mayor proporción que los hombres.
- **Clase**: Los pasajeros de primera clase tuvieron mejores probabilidades.
- **Edad**: La distribución por edad muestra diferencias moderadas, especialmente en niños.

Estas relaciones ayudan a identificar predictores clave para modelos posteriores.


##### 📌 Matriz de correlación

```python
plt.figure(figsize=(10,6))
numeric_df = df.select_dtypes(include=["int64", "float64"])
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Matriz de correlación (solo variables numéricas)")
plt.show()
```

### 🔗 Matriz de correlación

La matriz de correlación muestra qué variables numéricas están más relacionadas.
En este dataset:

- `Survived` correlaciona negativamente con `Pclass`.
- `Fare` correlaciona positivamente con `Survived`.

Esto sugiere un efecto importante de la clase social en la probabilidad de supervivencia.
