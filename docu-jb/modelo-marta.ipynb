# MODELO RANDOM FOREST

Este código importa las librerías esenciales para un proceso de Machine Learning. pandas permite cargar y manipular datos; train_test_split divide el dataset en entrenamiento y prueba; StandardScaler normaliza las variables numéricas; y RandomForestClassifier entrena un modelo basado en múltiples árboles de decisión. Finalmente, confusion_matrix y classification_report se utilizan para evaluar el rendimiento del modelo con métricas de clasificación.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, classification_report
```

## 1. Cargar datos (ejemplo)

Este fragmento de código hace lo siguiente:

Importa la librería pandas, que se utiliza para trabajar con datos en estructuras tipo DataFrame.

Usa pd.read_csv() para cargar un archivo CSV ubicado en ../datasets/Titanic-Dataset.csv y almacenarlo en la variable df.

df ahora contiene todos los datos del dataset Titanic en formato tabular, listos para exploración, limpieza o análisis.

```python
import pandas as pd

# Cargar el archivo CSV
df = pd.read_csv('../datasets/Titanic-Dataset.csv')
```

## 2. Features y variable objetivo

Se seleccionan las columnas Age y Fare como variables predictoras X, y la columna Survived como variable objetivo y.

```python
# Seleccionamos las columnas "Age" y "Fare" como variables predictoras (features)
X = df[["Age", "Fare"]]
# Seleccionamos la columna "Survived" como variable objetivo (label)
y = df["Survived"]
```

## 3. Dividir en train / test

Se divide el dataset en conjuntos de entrenamiento y prueba, usando el 25% de los datos para prueba y asegurando reproducibilidad con random_state=42.

```python
# Dividimos los datos en conjunto de entrenamiento y prueba, el 25% de los datos se usan para prueba
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

## 4. Escalado (Random Forest no lo requiere, pero ayuda si hay mucha variación)

Se crea un StandardScaler para estandarizar las variables, ajustándolo con los datos de entrenamiento y aplicando la transformación tanto al conjunto de entrenamiento como al de prueba.

```python
# Creamos un objeto StandardScaler para estandarizar las variables
scaler = StandardScaler()
# Ajustamos el scaler con los datos de entrenamiento y transformamos esos datos
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## 5. Modelo Random Forest

Este bloque de código realiza el entrenamiento y la predicción con un modelo de Random Forest para clasificación:

1. Se crea un objeto RandomForestClassifier configurado con 200 árboles (n_estimators=200), sin límite de profundidad (max_depth=None) y una semilla fija (random_state=42) para asegurar reproducibilidad.

2. El modelo se entrena usando los datos de entrenamiento previamente escalados (X_train_scaled) y la variable objetivo correspondiente (y_train).

3. Una vez entrenado, el modelo se utiliza para predecir los valores de la variable objetivo sobre el conjunto de prueba escalado (X_test_scaled), generando las predicciones y_pred.

Este flujo permite evaluar cómo se comporta el modelo con datos que no ha visto durante el entrenamiento.

```python
# Creamos un modelo de Random Forest para clasificación
rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    random_state=42
)

# Entrenamos el modelo con los datos de entrenamiento ya escalados
rf.fit(X_train_scaled, y_train)
# Realizamos predicciones sobre el conjunto de prueba escalado
y_pred = rf.predict(X_test_scaled)
```

## 6. Métricas

Este bloque de código se encarga de evaluar el desempeño del modelo:

1. confusion_matrix(y_test, y_pred) genera la matriz de confusión, que muestra cómo se distribuyen las predicciones correctas e incorrectas entre las clases reales.

2. classification_report(y_test, y_pred) imprime métricas detalladas como precisión (precision), exhaustividad (recall), F1-score y soporte (número de muestras por clase), ofreciendo una visión completa de la calidad del modelo en el conjunto de prueba.

En conjunto, estas métricas permiten identificar tanto la exactitud general como el comportamiento del modelo en cada clase individual.

```python
# Imprimimos la matriz de confusión
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Imprimimos el classification report
print("\nClassification Report:")
print(classification_report(y_test,y_pred))
```
